# 📄 Plagiarism / Text Similarity Checker

## 🎯 Objective

Compare two documents or passages and estimate how similar their wording is using TF-IDF and Cosine Similarity.


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text1 = input("Enter the first document/text: ")
text2 = input("Enter the second document/text: ")

if not text1.strip() or not text2.strip():
    print("Please enter both texts.")
else:
    vectorizer = TfidfVectorizer(stop_words="english")
    vectors = vectorizer.fit_transform([text1, text2])

    similarity = cosine_similarity(vectors[0:1], vectors[1:2])[0][0]
    percentage = similarity * 100

    print("\n===== TEXT SIMILARITY / PLAGIARISM CHECK =====")
    print(f"Similarity Score: {similarity:.4f}")
    print(f"Similarity Percentage: {percentage:.2f}%")

    if percentage >= 80:
        print("Result: Very High Similarity")
    elif percentage >= 60:
        print("Result: High Similarity")
    elif percentage >= 40:
        print("Result: Moderate Similarity")
    elif percentage >= 20:
        print("Result: Low Similarity")
    else:
        print("Result: Very Low Similarity")

    print("\nNote: This is a basic similarity demonstration, not a complete plagiarism detection system.")
```

## 💻 Example

**Text 1:** `Artificial intelligence is transforming modern technology.`

**Text 2:** `Artificial intelligence is changing modern technology.`


## 🧠 Concepts Used

**TF-IDF** for text vectorization and **Cosine Similarity** for comparing documents.


## ✅ Conclusion

The project demonstrates how text similarity can be used as a basic approach to identify potentially similar documents.
